```python
import pygeog510
```

```python
i_map = pygeog510.ipyMap(height="500px", zoom=3)
```

```python
i_map.add_geojson(
    "https://github.com/opengeos/datasets/releases/download/us/us_southeast.geojson"
)

i_map.add_layer_control()

i_map
```

```python
i_map_wms = pygeog510.ipyMap(center=[40, -100], height="700px", zoom=5)

i_map_wms.add_wms_layer(
    url="https://nowcoast.noaa.gov/geoserver/observations/weather_radar/wms",
    layers="conus_base_reflectivity_mosaic",
    format="image/png",
    transparent=True,
    attribution="NOAA nowCOAST",
)

i_map_wms
```

```python
i_map_image = pygeog510.ipyMap(center=[25, -120], height="500px", zoom=4)

bounds = [[12, -128], [30, -100]]

i_map_image.add_image(
    image="https://raw.githubusercontent.com/opengeos/data/refs/heads/main/images/weather.png",
    bounds=bounds,
)

i_map_image
```

```python
i_map_video = pygeog510.ipyMap(center=[25, -120], height="500px", zoom=4)

bounds = [[18.0, -98.0], [31.0, -82.0]]

i_map_video.add_video(
    video="https://samplelib.com/mp4/sample-10s-720p.mp4", bounds=bounds
)

i_map_video
```
