```python
# Assign initial variable as a String and output
geo_coord = "40.7128N, 74.0060W"
print(f"String: {geo_coord}")

# Extract lat and long from String and output each
lat_str, lon_str = geo_coord.split(", ")
print(f"\nLat Str: {lat_str}\nLon Str: {lon_str}")

# Convert Strings to Floats and strip dir indicator
lat_flt = float(lat_str[:-1])
lon_flt = float(lon_str[:-1])
print(
    f"Geo Coord (Float): {lat_flt:.4f}, {lon_flt:.4f}"
)  # Added formatting to maintain 4 places after decimal

# Format geocoord into a POINT Wkt String

wkt_point = f"POINT({lon_flt:.4f} {lat_flt:.4f})"  # Locking in 4 decimal places for readability after conversion
print(f"\nWKT: {wkt_point}")
```

```python
# Create list of coordinates

coordinates = [
    (34.0522, -118.2437),  # Los Angeles
    (-33.8688, 151.2093),  # Sydney
    (40.7128, -74.0060),  # New York
    (-23.5505, -46.6333),  # São Paulo
    (0.0000, 32.4467),  # Entebbe (Equator)
    (51.5074, -0.1278),  # London
    (-34.6037, -58.3816),  # Buenos Aires
    (35.6895, 139.6917),  # Tokyo
    (-1.2921, 36.8219),  # Nairobi
    (48.8566, 2.3522),  # Paris
    (38.6270, -90.1994),  # St. Louis
]

# Loop through coordinates list and check for Northern or Southern Hemisphere - output result

for lat, lon in coordinates:
    if lat > 0:
        print(f"The coordinate {lat:.4f}, {lon:.4f} is in the Northern Hemisphere.")
    elif lat < 0:
        print(f"The coordinate {lat:.4f}, {lon:.4f} is in the Southern Hemisphere.")
    else:
        print(f"The coordinate {lat:.4f}, {lon:.4f} is on the equator!")
```
